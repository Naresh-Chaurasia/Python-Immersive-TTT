# Complete Python Lambda Functions Tutorial

A comprehensive beginner-to-intermediate guide.

A **lambda** is a small anonymous function.

Syntax:
```python
lambda parameters: expression
```

````
square=lambda x:x*x
print(square(5))
````

## 2. Normal Function vs Lambda

````
def square(x):
    return x*x

square2=lambda x:x*x
print(square(4), square2(4))
````

## 3. Multiple Parameters

````
add=lambda a,b:a+b
print(add(10,20))
````

## 4. map()

````
nums=[1,2,3,4]
print(list(map(lambda x:x*x,nums)))
````

## 5. filter()

````
nums=range(10)
print(list(filter(lambda x:x%2==0,nums)))
````

## 6. sorted()

````
students=[{'name':'John','marks':80},{'name':'Amy','marks':95},{'name':'Bob','marks':70}]
print(sorted(students,key=lambda s:s['marks']))
````

## 7. max()

````
orders=[{'id':1,'amount':450},{'id':2,'amount':6500}]
print(max(orders,key=lambda o:o['amount']))
````

## 8. min()

````
print(min(orders,key=lambda o:o['amount']))
````

## 9. Real-world

````
employees=[{'name':'Alice','salary':90000},{'name':'Bob','salary':120000}]
print(sorted(employees,key=lambda e:e['salary'],reverse=True))
````

## When to Use Lambda
- Short one-line functions
- `map()`, `filter()`, `sorted()`, `min()`, `max()`
- Avoid for complex logic
